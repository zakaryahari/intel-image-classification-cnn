```python
from pathlib import Path

import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
DATA_PATH = Path(
    "../data/raw/seg_train"
)

IMG_SIZE = (150, 150)

BATCH_SIZE = 32

SEED = 42
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_PATH,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)
```

```python
val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_PATH,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)
```

```python
val_batches = len(val_ds)

test_size = val_batches // 2

test_ds = val_ds.take(
    test_size
)

val_ds = val_ds.skip(
    test_size
)
```

```python
class_names = train_ds.class_names

print(
    class_names
)
```

```python
for images, labels in train_ds.take(1):

    print(
        "Images shape:",
        images.shape
    )

    print(
        "Labels shape:",
        labels.shape
    )
```

```python
plt.figure(
    figsize=(10, 10)
)

for images, labels in train_ds.take(1):

    for i in range(9):

        plt.subplot(
            3,
            3,
            i + 1
        )

        plt.imshow(
            images[i].numpy().astype("uint8")
        )

        plt.title(
            class_names[
                labels[i]
            ]
        )

        plt.axis(
            "off"
        )

plt.show()
```
