```python
from pathlib import Path

import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
DATA_PATH = Path(
    "../data/raw/seg_train"
)

IMG_SIZE = (150, 150)

BATCH_SIZE = 32

SEED = 42
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_PATH,
    validation_split=0.2,
    subset="training",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)
```

```python
val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_PATH,
    validation_split=0.2,
    subset="validation",
    seed=SEED,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE
)
```

```python
val_batches = len(val_ds)

test_size = val_batches // 2

test_ds = val_ds.take(
    test_size
)

val_ds = val_ds.skip(
    test_size
)
```

```python
class_names = train_ds.class_names

print(
    class_names
)
```

```python
for images, labels in train_ds.take(1):

    print(
        "Images shape:",
        images.shape
    )

    print(
        "Labels shape:",
        labels.shape
    )
```

```python
plt.figure(
    figsize=(10, 10)
)

for images, labels in train_ds.take(1):

    for i in range(9):

        plt.subplot(
            3,
            3,
            i + 1
        )

        plt.imshow(
            images[i].numpy().astype("uint8")
        )

        plt.title(
            class_names[
                labels[i]
            ]
        )

        plt.axis(
            "off"
        )

plt.show()
```

```python
normalization_layer = tf.keras.layers.Rescaling(
    1./255
)
```

```python
for images, labels in train_ds.take(1):

    normalized_images = normalization_layer(
        images
    )

    print(
        "Before:",
        images[0].numpy().min(),
        images[0].numpy().max()
    )

    print(
        "After:",
        normalized_images[0].numpy().min(),
        normalized_images[0].numpy().max()
    )
```

```python
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip(
        "horizontal"
    ),

    tf.keras.layers.RandomRotation(
        0.1
    ),

    tf.keras.layers.RandomZoom(
        0.1
    )
])
```

```python
for images, labels in train_ds.take(1):

    image = images[0]

    plt.figure(
        figsize=(8, 8)
    )

    for i in range(9):

        augmented_image = data_augmentation(
            tf.expand_dims(
                image,
                0
            )
        )

        plt.subplot(
            3,
            3,
            i + 1
        )

        plt.imshow(
            augmented_image[0]
            .numpy()
            .astype("uint8")
        )

        plt.axis(
            "off"
        )

plt.show()
```

```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(
    buffer_size=AUTOTUNE
)

val_ds = val_ds.prefetch(
    buffer_size=AUTOTUNE
)

test_ds = test_ds.prefetch(
    buffer_size=AUTOTUNE
)
```
